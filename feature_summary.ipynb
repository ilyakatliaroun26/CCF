# Feature summary and descriptive statistics

```python
features = pd.read_csv("src/data/training/raw/merged_training_data.csv")
```

```python
features['mix__sd_od_history_to_kyc_ratio'] = features['od__history_length_days'] / features['sd__age_since_first_kyc']
```

```python
f_group = ['ab__', 'cb__', 'sd__', 'ep__', 'madrid__', 'mv__', 'ob__', 'ul__', 'dn__', 'mix']
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

f_group = ['ab__', 'cb__', 'sd__', 'ep__', 'madrid__', 'mv__', 'ob__', 'ul__', 'dn__', 'mix__']

for prefix in f_group:
    
    print("="*90)
    print(f"FEATURE GROUP: {prefix}")
    print("="*90)
    
    # Select columns belonging to this feature group
    group_cols = [col for col in features.columns if col.startswith(prefix)]
    
    if len(group_cols) == 0:
        print("No features found for this group.\n")
        continue
    
    df_group = features[group_cols]
    
    # ------------------------------------------------------------------
    # 1️⃣ Descriptive Statistics
    # ------------------------------------------------------------------
    print("\n--- Descriptive Statistics ---\n")
    display(df_group.describe(include='all').T)
    
    
    # ------------------------------------------------------------------
    # 2️⃣ Missing Values Plot
    # ------------------------------------------------------------------
    print("\n--- Missing Values Plot ---\n")
    
    missing_pct = df_group.isna().mean() * 100
    missing_pct = missing_pct.sort_values(ascending=False)
    
    plt.figure()
    plt.bar(missing_pct.index, missing_pct.values)
    plt.xticks(rotation=90)
    plt.ylabel("Missing %")
    plt.title(f"Missing Values - {prefix}")
    plt.tight_layout()
    plt.show()
    
    
    # ------------------------------------------------------------------
    # 3️⃣ Boxplots (Outliers – Numeric Only)
    # ------------------------------------------------------------------
    print("\n--- Boxplots (Outlier Detection) ---\n")
    
    numeric_cols = df_group.select_dtypes(include=np.number).columns
    
    for col in numeric_cols:
        plt.figure()
        plt.boxplot(df_group[col].dropna())
        plt.title(f"{col} - Boxplot")
        plt.ylabel(col)
        plt.tight_layout()
        plt.show()
    
    
    # ------------------------------------------------------------------
    # 4️⃣ Distributions
    # ------------------------------------------------------------------
    print("\n--- Distributions ---\n")
    
    # Numeric distributions (Histograms)
    for col in numeric_cols:
        plt.figure()
        plt.hist(df_group[col].dropna(), bins=30)
        plt.title(f"{col} - Distribution")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()
    
    
    # Categorical distributions (Top 10)
    categorical_cols = df_group.select_dtypes(exclude=np.number).columns
    
    for col in categorical_cols:
        value_counts = df_group[col].value_counts(dropna=False).head(10)
        
        plt.figure()
        plt.bar(value_counts.index.astype(str), value_counts.values)
        plt.xticks(rotation=90)
        plt.title(f"{col} - Top Categories")
        plt.ylabel("Count")
        plt.tight_layout()
        plt.show()
```
